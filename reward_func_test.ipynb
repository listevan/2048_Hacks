```python
from reward_func import get_reward
from Game_Sim.game import Game
import random
```

```python
g = Game()

g.display()

while True:
    prev = g._board.copy()

    move = random.randint(0, 3)
    successful, combined_values = g.move(move)
    next = g._board.copy()

    

    reward = get_reward(prev, move, next, g.check_win(), g.check_loss(), successful, combined_values)
    print('move: {}, reward: {}, win: {}, loss: {}'.format(move, reward, g.check_win(), g.check_loss()))

    g.display()

    x = input()
    if x == 'stop':
        break
```
